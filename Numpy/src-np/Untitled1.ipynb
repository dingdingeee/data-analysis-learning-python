```python
#asarray()#转化成ndarray
import numpy as np
a = [1,2,3,4,5,6,7,8,9]
c = np.asarray(a)
print(type(c))
```

```python
#frombuffer()
import numpy as np
h = b'hello numpy'#转成成ndarray:(bytes)转成成str
a = np.frombuffer(h,dtype = "S1")
print(a)#流
```

```python
#fromiter()：可以把迭代对象转换成ndarray
a = [1,2,3,4,5,6]
it = iter(a)#将列表转换成迭代器
print(it)

b = np.fromiter(it,dtype = int)#从迭代器当中去读取元素
print(b)
```

```python
#arrange(n)依次输入[0,n)，再用reshape()确定形状
import numpy as np
c = np.arange(24).reshape(4, 3, 2)
print(c)
```
