```python
import pandas as pd
import numpy as np

# 1. 空Series
s_empty = pd.Series()
print("空Series：\n", s_empty)

# 2. ndarray创建，自定义索引标签
arr = np.array(['Python', 'Java', 'C++'])
s_arr = pd.Series(arr, index=['A', 'B', 'C'])
print("\n自定义索引Series：\n", s_arr)

# 3. 字典创建 + 指定索引（索引一一对齐，缺失键自动补NaN）
dic = {'a': 10, 'b': 20, 'c': 30}
s_dic = pd.Series(dic, index=['c', 'b', 'a', 'd'])  # 'd' 在字典中不存在
print("\n字典创建+索引对齐：\n", s_dic)

# 4. 标量创建：所有元素填充同一个值
s_scalar = pd.Series(100, index=[0,1,2,3,4])
print("\n标量创建：\n", s_scalar)
```

```python
#访问Series数据    最好统一用.iloc()显式去写
#Series访问数据元素方式分为两种：一种是位置索引访问，另一种是索引标签访问
#1.位置索引访问
import pandas as pd
a = pd.Series([1,2,3,4,5],index = ['a','b','c','d','e'])
print(a)
#访问数据结构
print(a[-2:])#位置下标访问(可以配合切片技术)

print(a.iloc[-2:])
```

```python
#索引标签访问    最好统一用.loc()显式去写
import pandas as pd
a = pd.Series([1,2,3,4,5],index = ['a','b','c','d','e'])
print(a)
#索引标签访问
print(a['b'])#标签访问
#多标签访问
print(a[['a','b','c']])

print(a.loc[['a', 'b', 'c']])
```

```python
#最好显式去取这个值，前面的例子都可以用以下写法，减少歧义
#按标签取 -> .loc() | 按位置取 -> .iloc()
#i即为integer，按数字索引位置取，因为标签可能是数字、字符串、字典等等，所以注明可以减少歧义。
import numpy as np
import pandas as pd
arr = np.array([10, 20,30,40, 50])
print(type(arr))
s = pd.Series(arr, index=['a','b','c','d','e'])

# 位置索引：按顺序位置
print("位置索引取第2个元素：", s.iloc[1])
print("位置切片取后2个：\n", s.iloc[-2:])

# 标签索引：按index名称
print("标签索引取'b'：", s.loc['b'])
print("多标签批量选取：\n", s.loc[['a','c','e']])
```

```python
#isnull()&notnull()检测缺失值
import pandas as pd
a = pd.Series([1,2,None,4,5])
#print(pd.isnull(a)) #是空值返回True
print(pd.notnull(a)) #不是空值返回True
```

```python
"""
方法	功能说明
.head(n)	查看前 n 条数据，默认 n=5
.tail(n)	查看后 n 条数据，默认 n=5
.isnull()	缺失值返回 True，非缺失返回 False
.notnull()	非缺失返回 True，缺失返回 False（isnull 的反义）
"""

import numpy as np
import pandas as pd

a = [1, 2, None, 4, np.nan, 6]
s = pd.Series(a)

print("数据类型：", s.dtype)
print("元素总个数：", s.size)
print("转为numpy数组：", s.values)

print("\n缺失值布尔掩码：\n", s.isnull())
print("非缺失值布尔掩码：\n", s.notnull())

print("\n前3条数据：\n", s.head(3))
print("后2条数据：\n", s.tail(2))
```

```python
#有关创建DataFrame
import pandas as pd

# 1. 嵌套列表创建 + 指定列名、行索引
data_list = [['张三', 20, '男'],
             ['李四', 18, '女'],
             ['王五', 21, '男']]
df1 = pd.DataFrame(data_list, columns=['姓名','年龄','性别'], index=['s1','s2','s3'])
print("嵌套列表创建：\n", df1)

# 2. 字典嵌套列表创建（最常用）
data_dic = {
    '姓名': ['刘备','关羽','张飞'],
    '年龄': [38, 37, 35],
    '势力': ['蜀','蜀','蜀']
}
df2 = pd.DataFrame(data_dic, index=['A','B','C'])
print("\n字典嵌套列表创建：\n", df2)

# 3. 列表嵌套字典（按行），缺失列自动补NaN
data_row = [{'a':1, 'b':2}, {'a':4, 'b':5, 'c':6}]
df3 = pd.DataFrame(data_row, index=['行1','行2'])
print("\n列表嵌套字典创建：\n", df3)
```

```python
import pandas as pd

df = pd.DataFrame({
    'name': ['张三','李四','王五'],
    'age': [20, 18, 21]
}, index=[1,2,3])

# 1. 选取列
print("选取name列：\n", df['name'])
print("\n选取多列：\n", df[['name','age']])

# 2. 添加新列
df['gender'] = ['男','女','男']  # 列表逐行赋值
df['grade'] = 80  # 标量赋值，所有行填充同一个值
print("\n添加列后：\n", df)

# 3. 删除列
del df['grade']  # 直接原地删除
pop_col = df.pop('gender')  # 删除并返回被删列
print("\n删除列后：\n", df)
print("被弹出的列：\n", pop_col)

```

```python
import pandas as pd

df = pd.DataFrame([[1,2],[3,4]], columns=['a','b'], index=['x','y'])
print(df)
# 1. 选取行
print("按标签选x行：\n", df.loc['x'])
print("按位置选第1行：\n", df.iloc[1])
print("切片选前1行：\n", df[0:1])
print("按位置选取第y行的位置索引为1的元素：\n",df.loc['y'].iloc[1])

# 2. 追加新行
df_new = pd.DataFrame([[5,6],[7,8]], columns=['a','b'], index=['m','n'])
df = pd.concat([df, df_new], ignore_index = False)  # 必须接收返回值，注意如果忽略了原标签，那后续就不能用x\y\m\n做索引了，3.会报错
print("\n追加行后：\n", df)

# 3. 删除指定行
df = df.drop('x')  # 删除标签为x的行
print("\n删除x行后：\n", df)
```
