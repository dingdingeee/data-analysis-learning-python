```python
#CSV文件的读写
import pandas as pd

# 1. 构造测试数据
data = {
    '公司': ['百度', '腾讯', '网易', '酷狗'],
    '官网': ['www.baidu.com', 'www.tencent.com', 'www.163.com', 'www.kugou.com'],
    '成立年份': [2000, 1998, 1997, 2004]
}
df = pd.DataFrame(data)

# 2. 写入CSV：不导出行索引，使用utf-8-sig兼容Excel中文
df.to_csv('company.csv', index=False, encoding='utf-8-sig')

# 3. 读取CSV
df_read = pd.read_csv('company.csv', encoding='utf-8-sig')

# 4. 数据探查
print("前3行数据：")
print(df_read.head(3))
print("\n表格基本信息：")
print(df_read.info())

df1 = pd.read_csv('wwwe.csv')
df2 = pd.read_csv('www.csv')

# 5. 多工作表写入    pd.ExcelWriter()
with pd.ExcelWriter('multi_sheet1.xlsx', engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name = 'SheetA', index = False)
    df1.to_excel(writer, sheet_name = 'SheetB', index=False)
    df2.to_excel(writer, sheet_name = 'SheetC', index=False)
```

```python
#excel的多表写入
import pandas as pd

# 构造两组业务数据
df_sales = pd.DataFrame({'月份':['1月','2月','3月'], '销售额':[1200,1500,1800]})
df_user = pd.DataFrame({'地区':['华北','华东','华南'], '用户数':[2300,3100,2800]})

# 写入同一个Excel的不同工作表
with pd.ExcelWriter('report.xlsx', engine='openpyxl') as writer:
    df_sales.to_excel(writer, sheet_name='销售数据', index=False)
    df_user.to_excel(writer, sheet_name='用户数据', index=False)

# 读取指定工作表
df_read = pd.read_excel('report.xlsx', sheet_name='销售数据')
print("读取销售工作表：")
print(df_read)
```

```python
#json文件读取
import pandas as pd
#这个是什么保存方式保存数据的（列表嵌套字典）
data = [
        {"id":"01","name":"张三","age":18},
        {"id":"02","name":"王五","age":15},
        {"id":"03","name":"李四","age":22}]
#保存到数据结构中
df = pd.DataFrame(data)

#将DataFrame对象列表创建成json格式文件
df.to_json('www.json')
```

```python
#JSON文件的不同格式
import pandas as pd

df = pd.DataFrame({'id':[1,2,3], 'name':['张三','王五','李四'], 'age':[18,15,22]})

# 1. records格式：按行组织（最常用）
json_records = df.to_json(orient='records', force_ascii=False)
print("records格式：")
print(json_records)

# 2. columns格式：按列组织
json_cols = df.to_json(orient='columns', force_ascii=False)
print("\ncolumns格式：")
print(json_cols)

# 从JSON还原为DataFrame
df_back = pd.read_json(json_records, orient='records')
print("\n从records格式还原：")
print(df_back)
```
