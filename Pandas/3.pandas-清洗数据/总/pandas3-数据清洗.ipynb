```python
#多维查看缺失值
#判断缺失值
import pandas as pd
import numpy as np

a = pd.DataFrame({'A':[1,np.nan,3],
                  'B':[4,5,np.nan]})

#判断缺失值
print(a.isna())#isnull/isna,返回布尔类型，都是表示True是缺失值
```

```python
import pandas as pd
import numpy as np

a = pd.DataFrame({'A':[1,np.nan,3],
                  'B':[4,5,np.nan]})

#判断缺失值
print(a.fillna(0))#固定值填充
```

```python
#向前/向后填充:什么场景会用到：时间
import pandas as pd

a = pd.Series(['张三','李四','王五',None,'赵六'])

#标量替换填充
print(a.fillna(method = 'bfill'))#向前:ffill,向后填充:bfill
```

```python
#删除缺失值：通过dropna() 方法删除包含缺失值的行或列
#删除行：只要行当中有缺失值就删除，整行删除
import pandas as pd
import numpy as np

a = pd.DataFrame({'A':[1,np.nan,3],
                  'B':[4,5,np.nan],
                 'C':[4,5,7]})

df = a.dropna(subset = ['B', 'C']) #在指定的列里去判断是否有NaN，只能删除行
print(df)
print()
#判断缺失值
df1 = df.dropna(axis = 1)  #可以指定方向，默认是优先按行删除，多维也是
print(df1)
```

```python
#电商订单重复与缺失值处理
#业务：
#删除重复行（保留首次出现的数据行）
#用众数填充 “购买数量” 缺失值，用向前填充处理 “购买日期” 缺失值
#验证最终无缺失值和重复行

import pandas as pd
import numpy as np

data = {
    '订单编号': ['A001', 'A002', 'A003', 'A004', 'A001', 'A005', 'A006', 'A007', 'A002', 'A008'],
    '客户姓名': ['张三', '李四', '王五', '赵六', '张三', '孙七', '周八', '吴九', '李四', '郑十'],
    '商品名称': ['手机壳', '蓝牙耳机', '充电器', '数据线', '手机壳', '移动电源', '钢化膜', '机械键盘', '蓝牙耳机', '无线鼠标'],
    '购买数量': [2, np.nan, 1, 3, 2, np.nan, -5, 9999, np.nan, 1],
    '单价':    [29.9, 199.0, 59.0, 19.9, 29.9, 129.0, 15.0, 299.0, 199.0, 89.0],
    '购买日期': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-01',
                None, '2024-01-06', '2024-01-07', '2024-01-02', None],
    '支付方式': ['微信', '支付宝', '微信', '支付宝', '微信', '微信', '支付宝', '微信', '支付宝', '支付宝'],
}
df1 = pd.DataFrame(data)
df1.to_excel('comerce_test_dataset.xlsx')

#1.读取数据并删除重复数据行
df = pd.read_excel('comerce_test_dataset.xlsx')
#print(df)
df_dropped = df.drop_duplicates(keep = 'first')#first保留第一次出现的
print("去重后的数据：\n",df_dropped)

#填充缺失值
df_dropped['购买数量'] = df_dropped['购买数量'].fillna(df_dropped['购买数量'].mode().iloc(0))  # 众数填充
df_dropped['购买日期'] = df_dropped['购买日期'].ffill()  # 向前填充
print("填充后的数据：\n", df_dropped)

#验证
print("是否有缺失值：\n",df_dropped.isna().any())
print("是否有重复值：\n",df_dropped.duplicated().any())
```
