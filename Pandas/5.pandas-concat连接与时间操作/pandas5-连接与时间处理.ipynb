```python
#concat连接操作
import pandas as pd

#创建 DataFrame 1
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
}, index=[0, 1, 2])

# 创建 DataFrame 2
df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
}, index=[3, 4, 5])

# 创建 DataFrame 3（列不同，用于列拼接测试）
df3 = pd.DataFrame({
    'D': ['D0', 'D1', 'D2'],
    'E': ['E0', 'E1', 'E2'],
    'F': ['F0', 'F1', 'F2']
}, index=[0, 1, 2])

# 创建 DataFrame 4（索引重复，用于重置索引测试）
df4 = pd.DataFrame({
    'A': ['A6', 'A7'],
    'B': ['B6', 'B7'],
    'C': ['C6', 'C7']
}, index=[2, 3])
```

```python
result1 = pd.concat([df1,df2],ignore_index=True)#ignore_index=True重置索引
print('基础行拼接结果：')
print(result1)
```

```python
#行拼接（取列的交集，join='inner'）
result1 = pd.concat([df1,df4],join='inner')
print('基础行拼接结果：')
print(result1)
```

```python
#axis:0行方向，1列方向(将多个DataFrame按列拼接，要求索引尽量一致（不一致时用 NaN 填充）)

result1 = pd.concat([df1,df2],axis=1)#ignore_index=True重置索引
print('基础行拼接结果：')
print(result1)

result2 = pd.concat([df1,df4],axis=1,join='inner')#ignore_index=True重置索引
print('基础行拼接结果：')
print(result2)
```

```python
#Pandas日期时间格式化
import pandas as pd
# 1. 解析单个标准日期字符串
date_str1 = "2025-12-24"
a = pd.to_datetime(date_str1)
print(a)#日期时间对象
```

```python
# 2. 解析字符串列表（生成DatetimeIndex）
date_strs = ["2025-01-01", "2025-02-01", "2025-03-01"]
a = pd.to_datetime(date_strs)
print(a)
```

```python
# 3. 解析带时分秒的标准字符串
datetime_str = "2025-12-24 18:30:45"
a = pd.to_datetime(datetime_str)
print(a)
```

```python
#解析非标准格式字符串
# 1. 解析中文格式日期（2025年12月24日）
cn_date_str = "2025年12月24日"
a = pd.to_datetime(cn_date_str, format='%Y年%m月%d日')
print(a)#时间对象
```

```python
# 2. 解析日/月/年格式（24/12/2025）
eu_date_str = "24/12/2025"
ts_eu = pd.to_datetime(eu_date_str, format="%d/%m/%Y")
print("\n日/月/年格式解析：", ts_eu) 
```

```python
# 4. 解析带12小时制和上午/下午的时间（2025-12-24 06:30 PM）
ampm_datetime_str = "2025-12-24 06:30 PM"
ts_ampm = pd.to_datetime(ampm_datetime_str, format="%Y-%m-%d %I:%M %p")
print("\n12小时制时间解析：", ts_ampm)  # 输出：2025-12-24 18:30:00
```

```python
#Padans Timedelta时间差(直接指定时间单位的)
import pandas as pd
# 1. 创建单个时间差（1天3小时30分钟）
a = pd.Timedelta(days=1, hours=3, minutes=30)#days:天数，hours：小时，minutes：分钟
print(a)

# 2. 创建仅小时/秒的时间差
td2 = pd.Timedelta(hours=2, seconds=15)#hours：小时，seconds：秒
print(td2)
```

```python
# 3. 创建毫秒/微秒级时间差（高精度）
td3 = pd.Timedelta(milliseconds=500, microseconds=200)
print("时间差3：", td3) 
```
